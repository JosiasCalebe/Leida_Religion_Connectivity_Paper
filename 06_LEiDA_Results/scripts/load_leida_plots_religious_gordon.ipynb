## explore_state

```python
import matplotlib.pyplot as plt
import seaborn as sns 
from nilearn.plotting import plot_connectome
import re

def centroid2network(centroid):
    N_rois = centroid.size
    network_matrix = np.zeros((N_rois,N_rois))
    for node_idx,node_value in enumerate(centroid):
        if node_value>0:
            for node2_idx,node2_value in enumerate(centroid):
                if node2_value>0:
                    network_matrix[node_idx,node2_idx] = 1
                    network_matrix[node2_idx,node_idx] = 1
    return network_matrix


def get_significant_states(df, k):
    df_k = df[df['k'] == k].copy()

    # extrai o número do estado de "PL_state_X"
    df_k['state'] = df_k['variable'].str.extract(r'PL_state_(\d+)').astype(int)

    return set(df_k['state'].values)

def explore_state_2(centroid,rois_labels,occupancy,dwell_times,coords,state_number=None,darkstyle=False, 
    sig_occ_states=None,
    sig_dt_states=None):

    is_sig_occ = sig_occ_states and state_number in sig_occ_states
    is_sig_dt = sig_dt_states and state_number in sig_dt_states

    ax_positions = [['top'],
                    ['middle'],
                    ['low'],
                    ['left']
                    ]

    plt.ion()

    _, axd = plt.subplot_mosaic(
        ax_positions,
        figsize=(3, 16), 
        constrained_layout=False
        )


    #creating barplot with centroid values.
    centroid_ = pd.DataFrame(
        {'roi':rois_labels,
        'value':centroid}
        ).sort_values('value',ascending=False).reset_index()

    sns.barplot(
        x=centroid_.value,
        y=centroid_.roi,
        palette=['firebrick' if i>0 else 'royalblue' for i in centroid_.value],
        ax=axd['left']
        )

    if np.max(np.abs(centroid_.value))>0.15:
        axd['left'].set_xlim(-1.05,1.05)
    else:
        axd['left'].set_xlim(-0.11,0.11)

    #axd['left'].set_title('fMRI phase\nprojection '+r'into $V_{c}$',fontsize=10)
    #axd['left'].set_xlabel(' ')
    #axd['left'].set_ylabel('Brain regions',fontsize=16,labelpad=20)
    axd['left'].tick_params(labelsize=5,axis='y')
    axd['left'].axvline(0,color='black' if not darkstyle else 'white')
    axd['left'].set_yticklabels([])
    axd['left'].set_ylabel(None)
    axd['left'].set_xlabel(None)
    
    #creating brainplot
    network = centroid2network(centroid)
    edges_lw = {'linewidth':1,'color':'firebrick'}

    #nodes that don't belong to the
    #PL state are not shown
    if np.where(centroid>0)[0].size==0:
        sizes = 20
    else:
        sizes = [20 if roi>0 else 0 for roi in centroid]

    plot_connectome(
        network, 
        coords, 
        node_color='blue',
        node_size=sizes, 
        #edge_cmap=<matplotlib.colors.LinearSegmentedColormap object>, 
        edge_vmin=None, 
        edge_vmax=None, 
        edge_threshold=None, 
        output_file=None, 
        display_mode='z', 
        figure=_, 
        axes=axd['top'], 
        annotate=True, 
        black_bg=True if darkstyle else False, 
        alpha=0.1, 
        edge_kwargs=edges_lw, 
        node_kwargs=None, 
        colorbar=False
        )

    #creating boxplot with occupancies
    occupancy.columns = ['condition','value']

    axd['top'].set_title('FC State ' + str(state_number), fontweight='bold')

    str_length = [len(cond) for cond in np.unique(occupancy.condition)]
    N_conds = np.unique(occupancy.condition).size

    sns.boxplot(
        x=occupancy.condition,
        y=occupancy.value,
        ax=axd['middle'],
        color='black' if not darkstyle else 'white',
        width=.4,
        fliersize=.4,
        #linewidth=.8,
        #boxprops = dict(linewidth=3,color='black'),
        medianprops = dict(linewidth=1.5, color='black' if darkstyle else 'white')
        )
    axd['middle'].set_ylabel('Occupancy',fontsize=15)
    axd['middle'].set_xlabel('')
    axd['middle'].tick_params(
        labelsize=10,
        axis='x',
        labelrotation=0 if (N_conds<=2 and np.max(str_length)<5) else 45
        )
    sns.despine(ax=axd['middle'])

    #creating boxplot with dwell times
    dwell_times.columns = ['condition','value']
    N_conds = np.unique(dwell_times.condition).size
    sns.boxplot(
        x=dwell_times.condition,
        y=dwell_times.value,
        ax=axd['low'],
        color='black' if not darkstyle else 'white',
        width=.4,
        fliersize=.4,
        #linewidth=.8,
        #boxprops = dict(linewidth=3, color='black'),
        medianprops = dict(linewidth=1.5, color='black' if darkstyle else 'white')
        )
    axd['low'].set_ylabel('Dwell time',fontsize=15)
    axd['low'].set_xlabel('')
    axd['low'].tick_params(
        labelsize=10,
        axis='x',
        labelrotation=0 if (N_conds<=2 and np.max(str_length)<5) else 45
        )
    sns.despine(ax=axd['low'])

    #plt.tight_layout(w_pad=0.7,h_pad=10)
    plt.subplots_adjust(hspace=0.4)
    plt.show()
```

```python
import os

def get_significant_states(df, k):
    
    if not isinstance(df, pd.DataFrame):
        return set()

    if df.empty:
        return set()
    
    df_k = df[df['k'] == k].copy()

    # extrai o número do estado de "PL_state_X"
    df_k['variable'] = df_k['variable'].str.extract(r'PL_state_(\d+)').astype(int)

    return set(df_k['variable'].values)
  
def plot_states_grid(k, ld, n_cols, save_path, darkstyle=False):

    sig_occ_states = get_significant_states(ld.significant_states(metric='occupancies'), k)
    sig_dt_states = get_significant_states(ld.significant_states(metric='dwell_times'), k)
    centroids = ld.load_centroids(k=k).values
    n_states = centroids.shape[0]

    fig, axes = plt.subplots(
        4, n_states,
        figsize=(3*n_states, 16),
        constrained_layout=False
    )

    for i in range(n_states):

        is_sig_occ = sig_occ_states and (i+1) in sig_occ_states
        is_sig_dt = sig_dt_states and (i+1) in sig_dt_states

        centroid = centroids[i, :]
        state = i + 1

        occ = ld.occupancies(k=k)[['condition', f'PL_state_{state}']]
        dt = ld.dwell_times(k=k)[['condition', f'PL_state_{state}']]

        # reaproveita axes por coluna
        axd = {
            'top': axes[0, i],
            'middle': axes[1, i],
            'low': axes[2, i],
            'left': axes[3, i],
        }

        # 🔥 reaproveita código interno (copiando o corpo principal)
        centroid_ = pd.DataFrame(
            {'roi': ld.rois_labels, 'value': centroid}
        ).sort_values('value', ascending=False).reset_index()

        sns.barplot(
            x=centroid_.value,
            y=centroid_.roi,
            palette=['firebrick' if v > 0 else 'royalblue' for v in centroid_.value],
            ax=axd['left']
        )

        axd['left'].axvline(0, color='black')
        axd['left'].set_yticklabels([])
        axd['left'].set_xlabel(None)
        axd['left'].set_ylabel(None)

        # conectoma
        network = centroid2network(centroid)
        sizes = [20 if roi > 0 else 0 for roi in centroid]

        plot_connectome(
            network,
            ld.rois_coordinates_,
            node_color='blue',
            node_size=sizes,
            display_mode='z',
            figure=fig,
            axes=axd['top'],
            annotate=False,
            black_bg=darkstyle,
            alpha=0.1,
            edge_kwargs={'linewidth':1,'color':'firebrick'},
            colorbar=False
        )

        axd['top'].set_title(f'State {state}')

        # occupancy
        occ.columns = ['condition','value']
        sns.boxplot(
            x=occ.condition,
            y=occ.value,
            ax=axd['middle'],
            color='black'
        )

        # dwell time
        dt.columns = ['condition','value']
        sns.boxplot(
            x=dt.condition,
            y=dt.value,
            ax=axd['low'],
            color='black'
        )

        axd['middle'].set_xlabel('')
        axd['middle'].set_ylabel('')
        axd['middle'].set_title('Occupancies', fontsize=12, pad=10)
        if is_sig_occ:
            axd['middle'].set_facecolor('#fff3cd')
              # amarelo claro
        axd['low'].set_xlabel('')
        axd['low'].set_ylabel('')
        axd['low'].set_title('Dwell Times', fontsize=12, pad=10)
        if is_sig_dt:
            axd['low'].set_facecolor('#d1ecf1')  # azul claro
    
    plt.subplots_adjust(wspace=0.4, hspace=0.4)
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.close()
```

```python
def plot_k10_split(ld):

    centroids = ld.load_centroids(k=10).values

    for part in range(2):
        start = part * 5
        end = start + 5

        fig, axes = plt.subplots(4, 5, figsize=(15,16))

        for i, state_idx in enumerate(range(start, end)):
            centroid = centroids[state_idx, :]
            state = state_idx + 1

            occ = ld.occupancies(k=10)[['condition', f'PL_state_{state}']]
            dt = ld.dwell_times(k=10)[['condition', f'PL_state_{state}']]

            axd = {
                'top': axes[0, i],
                'middle': axes[1, i],
                'low': axes[2, i],
                'left': axes[3, i],
            }

            # reaproveita MESMO código do grid acima (copiar bloco)

            centroid_ = pd.DataFrame(
                {'roi': ld.rois_labels, 'value': centroid}
            ).sort_values('value', ascending=False).reset_index()

            sns.barplot(
                x=centroid_.value,
                y=centroid_.roi,
                palette=['firebrick' if v > 0 else 'royalblue' for v in centroid_.value],
                ax=axd['left']
            )

            axd['left'].axvline(0, color='black')
            axd['left'].set_yticklabels([])

            network = centroid2network(centroid)
            sizes = [20 if roi > 0 else 0 for roi in centroid]

            plot_connectome(
                network,
                ld.rois_coordinates_,
                node_color='blue',
                node_size=sizes,
                display_mode='z',
                figure=fig,
                axes=axd['top'],
                annotate=False,
                black_bg=False,
                alpha=0.1,
                edge_kwargs={'linewidth':1,'color':'firebrick'},
                colorbar=False
            )

            axd['top'].set_title(f'State {state}')

            occ.columns = ['condition','value']
            sns.boxplot(x=occ.condition, y=occ.value, ax=axd['middle'], color='black')

            dt.columns = ['condition','value']
            sns.boxplot(x=dt.condition, y=dt.value, ax=axd['low'], color='black')

            axd['middle'].set_xlabel('')
            axd['middle'].set_ylabel('')
            axd['middle'].set_title('Occupancies', fontsize=12, pad=10)

            axd['low'].set_xlabel('')
            axd['low'].set_ylabel('')
            axd['low'].set_title('Dwell Times', fontsize=12, pad=10)

        plt.savefig(f'k10_part_{part+1}.png', dpi=300, bbox_inches='tight')
        plt.close()

plot_k10_split(ld)
```
